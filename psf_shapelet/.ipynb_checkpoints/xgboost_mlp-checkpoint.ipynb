# XGBoost and MLP: PSF Moments → Zernike Coefficients

Trains two baseline models to predict Zernike wavefront coefficients at 4 AOS
corners from per-raft PSF moments, following `xgboost_mlp_guide.txt`.

- **Input** X ∈ ℝ^{N×168} (21 rafts × 8 PSF moments, flattened)
- **Output** Y ∈ ℝ^{N×32} (4 corners × 8 Zernike modes Z4–Z11)

Loads preprocessed `dataset.npz` and `scalers.pkl` produced by `prepare_dataset.ipynb`.

**Workflow**
1. Load & flatten arrays, sanity checks
2. Mean-predictor baseline
3. XGBoost (one model per output, early stopping)
4. MLP (shared representation, AdamW + cosine LR)
5. Comparison: RMSE / Pearson r per mode and per corner

```python
import numpy as np
import pickle
import matplotlib.pyplot as plt
import xgboost as xgb
import torch
import torch.nn as nn
import torch.nn.functional as F
from torch.utils.data import DataLoader, TensorDataset

device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print(f'Using device: {device}')
```

## Config

```python
DATASET_PATH = 'data/dataset.npz'
SCALERS_PATH = 'data/scalers.pkl'

ZERNIKE_MODES = ['z4', 'z5', 'z6', 'z7', 'z8', 'z9', 'z10', 'z11']
CORNER_NAMES  = ['SW', 'SE', 'NW', 'NE']
N_ZERNIKE     = len(ZERNIKE_MODES)   # 8
N_CORNERS     = len(CORNER_NAMES)    # 4
N_RAFTS       = 21
N_MOMENTS     = 8

N_INPUT  = N_RAFTS * N_MOMENTS   # 168
N_OUTPUT = N_CORNERS * N_ZERNIKE  # 32

# XGBoost defaults
XGB_PARAMS = dict(
    n_estimators      = 500,
    learning_rate     = 0.05,
    max_depth         = 5,
    min_child_weight  = 3,
    subsample         = 0.8,
    colsample_bytree  = 0.8,
    reg_lambda        = 1.0,
    reg_alpha         = 0.0,
    objective         = 'reg:squarederror',
    tree_method       = 'hist',
    n_jobs            = -1,
    random_state      = 42,
)
EARLY_STOPPING_ROUNDS = 40
```

## Load dataset

```python
data = np.load(DATASET_PATH)
X_scaled  = data['X_scaled']   # [N, 21, 8] standardised
Y_scaled  = data['Y_scaled']   # [N, 4, 8]  standardised
X_raw     = data['X_raw']      # [N, 21, 8] physical
Y_raw     = data['Y_raw']      # [N, 4, 8]  physical
idx_train = data['idx_train']
idx_test  = data['idx_test']

with open(SCALERS_PATH, 'rb') as f:
    scalers = pickle.load(f)
y_scaler = scalers['y_scaler']

N = len(X_scaled)
print(f'Total: {N}  Train: {len(idx_train)}  Test: {len(idx_test)}')

# Flatten to [N, 168] and [N, 32]
X_flat = X_scaled.reshape(N, N_INPUT)
Y_flat = Y_scaled.reshape(N, N_OUTPUT)

Xtr, Xte = X_flat[idx_train], X_flat[idx_test]
Ytr, Yte = Y_flat[idx_train], Y_flat[idx_test]

print(f'X_flat shape: {X_flat.shape}  Y_flat shape: {Y_flat.shape}')
```

## Sanity checks

```python
assert not np.any(np.isnan(X_flat)), 'NaN in X'
assert not np.any(np.isnan(Y_flat)), 'NaN in Y'
assert X_flat.shape[1] == N_INPUT,   f'Expected {N_INPUT} input features'
assert Y_flat.shape[1] == N_OUTPUT,  f'Expected {N_OUTPUT} output targets'

print('Sanity checks passed')
print(f'X_flat: mean={X_flat.mean():.3f}  std={X_flat.std():.3f}')
print(f'Y_flat: mean={Y_flat.mean():.3f}  std={Y_flat.std():.3f}')

# Target std per output (in standardised space — should be ~1 on train)
print('\nTarget std per Zernike mode (train, standardised):')
Ytr_3d = Ytr.reshape(-1, N_CORNERS, N_ZERNIKE)
for k, m in enumerate(ZERNIKE_MODES):
    print(f'  {m}: {Ytr_3d[:, :, k].std():.3f}')
```

## Helper: evaluate predictions

```python
def unscale_Y(Y_s):
    """Inverse-transform standardised [N, 32] → physical [N, 32]."""
    return y_scaler.inverse_transform(Y_s.reshape(-1, N_ZERNIKE)).reshape(Y_s.shape[0], N_CORNERS, N_ZERNIKE)


def evaluate(Y_pred_s, Y_true_s, label=''):
    """Compute and print RMSE/Pearson r per mode and per corner (physical units)."""
    Y_pred = unscale_Y(Y_pred_s)   # [N, 4, 8]
    Y_true = unscale_Y(Y_true_s)

    rmse_mode   = np.sqrt(np.mean((Y_pred - Y_true)**2, axis=(0, 1)))  # [8]
    rmse_corner = np.sqrt(np.mean((Y_pred - Y_true)**2, axis=(0, 2)))  # [4]
    macro_rmse  = rmse_mode.mean()

    print(f'\n=== {label} ===')
    print(f'Macro RMSE [μm]: {macro_rmse:.4f}')
    print('RMSE per Zernike mode [μm]:')
    for m, r in zip(ZERNIKE_MODES, rmse_mode):
        print(f'  {m}: {r:.4f}')
    print('RMSE per corner [μm]:')
    for c, r in zip(CORNER_NAMES, rmse_corner):
        print(f'  {c}: {r:.4f}')
    print('Pearson r per Zernike mode:')
    for k, m in enumerate(ZERNIKE_MODES):
        p, t = Y_pred[:, :, k].ravel(), Y_true[:, :, k].ravel()
        print(f'  {m}: {np.corrcoef(p, t)[0,1]:.3f}')

    return dict(rmse_mode=rmse_mode, rmse_corner=rmse_corner, macro_rmse=macro_rmse,
                Y_pred=Y_pred, Y_true=Y_true)
```

## Baseline: mean predictor

```python
train_mean = Ytr.mean(axis=0, keepdims=True)   # [1, 32]
Y_mean_pred = np.repeat(train_mean, len(Yte), axis=0)
res_mean = evaluate(Y_mean_pred, Yte, 'Mean predictor (test)')
```

## XGBoost: one model per output

```python
# 70/15/15 split: carve a validation set out of idx_train
val_frac = 15 / 85   # so that final val ≈ 15% of full N
rng = np.random.default_rng(SEED)
perm = rng.permutation(len(idx_train))
n_val = int(len(idx_train) * val_frac)
val_local, train_local = perm[:n_val], perm[n_val:]

Xtr_xgb  = Xtr[train_local]
Ytr_xgb  = Ytr[train_local]
Xval_xgb = Xtr[val_local]
Yval_xgb = Ytr[val_local]

print(f'XGBoost  train: {Xtr_xgb.shape[0]}  val: {Xval_xgb.shape[0]}  test: {Xte.shape[0]}')
```

```python
XGB_PARAMS
```

```python
xgb_models = []
xgb_test_preds = np.zeros((len(Xte), N_OUTPUT), dtype=np.float32)

for d in range(N_OUTPUT):
    corner_idx = d // N_ZERNIKE
    mode_idx   = d %  N_ZERNIKE
    label      = f'{CORNER_NAMES[corner_idx]}_{ZERNIKE_MODES[mode_idx]}'

    model = xgb.XGBRegressor(**XGB_PARAMS, early_stopping_rounds=EARLY_STOPPING_ROUNDS)
    model.fit(
        Xtr_xgb, Ytr_xgb[:, d],
        eval_set=[(Xval_xgb, Yval_xgb[:, d])],
        verbose=False,
    )
    xgb_test_preds[:, d] = model.predict(Xte)
    xgb_models.append(model)

    if (d + 1) % 8 == 0:
        print(f'Trained outputs 1–{d+1}/{N_OUTPUT}')

print('XGBoost training complete')
```

```python
res_xgb = evaluate(xgb_test_preds, Yte, 'XGBoost (test)')
```

```python

# Predicted vs true: all 32 outputs
# Layout: 4 rows × 8 cols — corners at focal-plane positions
#   NW (top-left)  | NE (top-right)
#   SW (bot-left)  | SE (bot-right)
from matplotlib.colors import LogNorm
from matplotlib.lines import Line2D

Y_pred_phys = unscale_Y(xgb_test_preds)
Y_true_phys = unscale_Y(Yte)

BLOCK_POS = {'SW': (1, 0), 'SE': (1, 1), 'NW': (0, 0), 'NE': (0, 1)}

n_rows, n_cols = 4, 8
fig, axes = plt.subplots(n_rows, n_cols, figsize=(n_cols * 3.5, n_rows * 3.5))

for c, corner in enumerate(CORNER_NAMES):
    br, bc = BLOCK_POS[corner]
    for k, mode in enumerate(ZERNIKE_MODES):
        row = br * 2 + k // 4
        col = bc * 4 + k % 4
        ax  = axes[row, col]
        t = Y_true_phys[:, c, k]
        p = Y_pred_phys[:, c, k]
        vmin = min(t.min(), p.min()); vmax = max(t.max(), p.max())
        margin = (vmax - vmin) * 0.05
        lim = (vmin - margin, vmax + margin)
        ax.hist2d(t, p, bins=50, range=[[-1, 1], [-1, 1]], cmap='magma', norm=LogNorm())
        ax.plot(lim, lim, 'b--', lw=0.7, alpha=0.7)
        ax.set_aspect('equal', adjustable='box')
        ax.set_title(f'{corner} · {mode}', fontsize=11)
        ax.tick_params(labelsize=9)
        if col == 0 or col == 4:
            ax.set_ylabel('Pred [μm]', fontsize=10)
        if row == n_rows - 1:
            ax.set_xlabel('True [μm]', fontsize=10)

for corner, (br, bc) in BLOCK_POS.items():
    cx = (bc * 4 + 1.5) / n_cols
    cy = 1.0 - br * 0.5 - 0.02
    fig.text(cx, cy, corner, ha='center', va='bottom', fontsize=14, fontweight='bold',
             transform=fig.transFigure)

fig.suptitle('XGBoost — one model per output\nPredicted vs True (test set)', fontsize=16, y=1.03)
plt.tight_layout()

# Divider lines between corner blocks
pos_l = axes[0, 3].get_position(); pos_r = axes[0, 4].get_position()
pos_t = axes[1, 0].get_position(); pos_b = axes[2, 0].get_position()
x_div = (pos_l.x1 + pos_r.x0) / 2
y_div = (pos_t.y0 + pos_b.y1) / 2
fig.add_artist(Line2D([x_div, x_div], [0.01, 0.99], transform=fig.transFigure,
                      color='white', linewidth=3, zorder=10))
fig.add_artist(Line2D([0.01, 0.99], [y_div, y_div], transform=fig.transFigure,
                      color='white', linewidth=3, zorder=10))

plt.savefig('fig/xgb_pred_vs_true_all32.png', dpi=150, bbox_inches='tight')
plt.show()
```

## XGBoost hyperparameter search (30 random trials)

Searches over 7 key parameters.  Each trial trains 32 models (one per output) with early stopping
and scores by **mean validation RMSE** across all 32 outputs.

Search space:

| Parameter | Values |
|---|---|
| `learning_rate` | 0.01, 0.03, 0.05, 0.10 |
| `max_depth` | 3, 4, 5, 6, 7 |
| `min_child_weight` | 1, 3, 5, 7 |
| `subsample` | 0.6, 0.7, 0.8, 0.9, 1.0 |
| `colsample_bytree` | 0.6, 0.7, 0.8, 0.9, 1.0 |
| `reg_lambda` | 0.5, 1.0, 2.0, 5.0 |
| `reg_alpha` | 0.0, 0.1, 0.5, 1.0 |

```python
import time

XGB_SEARCH_SPACE = {
    'learning_rate':    [0.01, 0.03, 0.05, 0.10],
    'max_depth':        [3, 4, 5, 6, 7],
    'min_child_weight': [1, 3, 5, 7],
    'subsample':        [0.6, 0.7, 0.8, 0.9, 1.0],
    'colsample_bytree': [0.6, 0.7, 0.8, 0.9, 1.0],
    'reg_lambda':       [0.5, 1.0, 2.0, 5.0],
    'reg_alpha':        [0.0, 0.1, 0.5, 1.0],
}
XGB_N_TRIALS    = 30
XGB_SEARCH_PAT  = 20   # early stopping patience during search (faster)
XGB_FIXED = dict(
    n_estimators = 600,
    objective    = 'reg:squarederror',
    tree_method  = 'hist',
    n_jobs       = -1,
    random_state = 42,
)

rng_xgb = np.random.default_rng(7)

def sample_xgb_cfg():
    return {k: rng_xgb.choice(v).item() for k, v in XGB_SEARCH_SPACE.items()}

def score_xgb_cfg(cfg, patience=XGB_SEARCH_PAT):
    """Train 32 XGB models with cfg; return (mean_val_rmse, per_output_rmse)."""
    rmse = np.zeros(N_OUTPUT)
    for d in range(N_OUTPUT):
        m = xgb.XGBRegressor(**XGB_FIXED, **cfg,
                             early_stopping_rounds=patience)
        m.fit(Xtr_xgb, Ytr_xgb[:, d],
              eval_set=[(Xval_xgb, Yval_xgb[:, d])],
              verbose=False)
        rmse[d] = np.sqrt(m.best_score)
    return rmse.mean(), rmse

xgb_trials = []
t0 = time.time()
print(f"{'Trial':>5}  {'val_RMSE':>9}  {'lr':>6}  {'depth':>5}  {'mcw':>4}  "
      f"{'sub':>5}  {'col':>5}  {'λ':>5}  {'α':>5}  {'time':>6}")
print('─' * 75)

for trial in range(XGB_N_TRIALS):
    cfg   = sample_xgb_cfg()
    score, per_out = score_xgb_cfg(cfg)
    xgb_trials.append({'cfg': cfg, 'val_rmse': score, 'per_output': per_out})
    elapsed = time.time() - t0
    marker = ' ◀ best' if score == min(t['val_rmse'] for t in xgb_trials) else ''
    print(f"{trial+1:5d}  {score:9.5f}  {cfg['learning_rate']:6.3f}  "
          f"{cfg['max_depth']:5d}  {cfg['min_child_weight']:4d}  "
          f"{cfg['subsample']:5.2f}  {cfg['colsample_bytree']:5.2f}  "
          f"{cfg['reg_lambda']:5.2f}  {cfg['reg_alpha']:5.2f}  "
          f"{elapsed:5.0f}s{marker}")

xgb_trials.sort(key=lambda x: x['val_rmse'])
best_xgb_cfg = xgb_trials[0]['cfg']
print(f'\nBest val_RMSE : {xgb_trials[0]["val_rmse"]:.5f}')
print(f'Default val_RMSE comparison — run evaluate() on val set above to compare.')
print('Best config:')
for k, v in best_xgb_cfg.items():
    print(f'  {k:<20s} = {v}')
```

```python
# Search summary: top-10 trials + visualisation
print(f"Top 10 trials:")
print(f"{'Rank':>4}  {'val_RMSE':>9}  {'lr':>6}  {'depth':>5}  {'mcw':>4}  "
      f"{'sub':>5}  {'col':>5}  {'λ':>5}  {'α':>5}")
print('─' * 62)
for i, t in enumerate(xgb_trials[:10]):
    c = t['cfg']
    print(f"{i+1:4d}  {t['val_rmse']:9.5f}  {c['learning_rate']:6.3f}  "
          f"{c['max_depth']:5d}  {c['min_child_weight']:4d}  "
          f"{c['subsample']:5.2f}  {c['colsample_bytree']:5.2f}  "
          f"{c['reg_lambda']:5.2f}  {c['reg_alpha']:5.2f}")

# Plot: trial score vs rank + per-parameter marginal distributions
scores_sorted = [t['val_rmse'] for t in xgb_trials]  # already sorted best→worst

fig, axes = plt.subplots(2, 4, figsize=(16, 7))
fig.suptitle('XGBoost Hyperparameter Search — 30 Trials', fontsize=13, fontweight='bold')

# Top row: val_RMSE vs trial rank
ax = axes[0, 0]
colors = ['#1565c0' if i < 5 else '#90a4ae' for i in range(XGB_N_TRIALS)]
ax.bar(range(1, XGB_N_TRIALS+1), scores_sorted, color=colors)
ax.axhline(scores_sorted[0], color='#c0392b', lw=1.5, ls='--', label=f'best={scores_sorted[0]:.4f}')
ax.set_xlabel('Rank'); ax.set_ylabel('Mean val RMSE')
ax.set_title('Score by rank'); ax.legend(fontsize=8)

# Remaining: scatter of each HP vs score (coloured by top-5 vs rest)
params = list(XGB_SEARCH_SPACE.keys())
for pi, param in enumerate(params):
    row, col = divmod(pi + 1, 4)
    ax = axes[row, col]
    vals  = [t['cfg'][param] for t in xgb_trials]
    rmses = [t['val_rmse']   for t in xgb_trials]
    top5  = [i < 5 for i in range(XGB_N_TRIALS)]
    ax.scatter([v for v,f in zip(vals,top5) if not f],
               [r for r,f in zip(rmses,top5) if not f],
               s=30, alpha=0.6, color='#90a4ae', label='other')
    ax.scatter([v for v,f in zip(vals,top5) if f],
               [r for r,f in zip(rmses,top5) if f],
               s=60, color='#1565c0', zorder=3, label='top-5', marker='D')
    ax.set_xlabel(param, fontsize=9); ax.set_ylabel('val RMSE', fontsize=8)
    ax.set_title(param, fontsize=9)
    ax.legend(fontsize=7)

plt.tight_layout()
plt.savefig('fig/xgb_search.png', dpi=150, bbox_inches='tight')
plt.show()
print('Saved fig/xgb_search.png')
```

```python
# Retrain best config on full train set (train_local + val_local), test on held-out test
RETRAIN_PARAMS = dict(**XGB_FIXED, **best_xgb_cfg)

xgb_models_tuned = []
xgb_tuned_preds  = np.zeros((len(Xte), N_OUTPUT), dtype=np.float32)
idx_full_train   = np.concatenate([train_local, val_local])  # all train rows

for d in range(N_OUTPUT):
    m = xgb.XGBRegressor(**RETRAIN_PARAMS, early_stopping_rounds=40)
    m.fit(
        Xtr[idx_full_train], Ytr[idx_full_train, d],
        eval_set=[(Xval_xgb, Yval_xgb[:, d])],   # val set for early stopping
        verbose=False,
    )
    xgb_tuned_preds[:, d] = m.predict(Xte)
    xgb_models_tuned.append(m)
    if (d + 1) % 8 == 0:
        print(f'Retrained {d+1}/{N_OUTPUT}')

print('Retraining complete')
res_xgb_tuned = evaluate(xgb_tuned_preds, Yte, 'XGBoost tuned (test)')
```

```python
# Default vs tuned XGBoost comparison
labels = ZERNIKE_MODES
x = np.arange(N_ZERNIKE)
w = 0.35

fig, axes = plt.subplots(1, 2, figsize=(14, 4.5))
fig.suptitle('XGBoost Default vs Tuned — Test RMSE', fontsize=12, fontweight='bold')

# Per-mode RMSE
ax = axes[0]
ax.bar(x - w/2, res_xgb['rmse_mode'],       w, label='Default', color='#90a4ae')
ax.bar(x + w/2, res_xgb_tuned['rmse_mode'], w, label='Tuned',   color='#1565c0')
ax.set_xticks(x); ax.set_xticklabels([m.upper() for m in ZERNIKE_MODES])
ax.set_ylabel('RMSE (standardised)'); ax.set_title('Per Zernike mode')
ax.legend()

# Per-corner RMSE
ax = axes[1]
xc = np.arange(N_CORNERS)
ax.bar(xc - w/2, res_xgb['rmse_corner'],       w, label='Default', color='#90a4ae')
ax.bar(xc + w/2, res_xgb_tuned['rmse_corner'], w, label='Tuned',   color='#1565c0')
ax.set_xticks(xc); ax.set_xticklabels(CORNER_NAMES)
ax.set_ylabel('RMSE (standardised)'); ax.set_title('Per AOS corner')
ax.legend()

for ax in axes:
    ax.set_ylim(0)
    ax.grid(axis='y', alpha=0.4)

plt.tight_layout()
plt.savefig('fig/xgb_default_vs_tuned.png', dpi=150, bbox_inches='tight')
plt.show()
print('Saved fig/xgb_default_vs_tuned.png')

# print(f"\nOverall RMSE — default: {res_xgb['rmse_overall']:.5f}  "
#       f"tuned: {res_xgb_tuned['rmse_overall']:.5f}  "
#       f"(Δ = {res_xgb['rmse_overall'] - res_xgb_tuned['rmse_overall']:+.5f})")
```

## MLP: shared representation

```python

# MLP training
MLP_BATCH_SIZE = 64
MLP_EPOCHS     = 1000
MLP_LR         = 1e-3
MLP_WD         = 1e-4
MLP_PATIENCE   = 25
MLP_DROPOUT    = 0.1

SEED = 42
torch.manual_seed(SEED)
np.random.seed(SEED)
```

```python
class PSFToZernikeMLP(nn.Module):
    def __init__(self, in_dim=N_INPUT, out_dim=N_OUTPUT, dropout=MLP_DROPOUT):
        super().__init__()
        self.net = nn.Sequential(
            nn.Linear(in_dim, 256),
            nn.GELU(),
            nn.LayerNorm(256),
            nn.Dropout(dropout),

            nn.Linear(256, 256),
            nn.GELU(),
            nn.LayerNorm(256),
            nn.Dropout(dropout),

            nn.Linear(256, 128),
            nn.GELU(),
            nn.LayerNorm(128),
            nn.Dropout(dropout),

            nn.Linear(128, out_dim),
        )

    def forward(self, x):
        return self.net(x)


mlp = PSFToZernikeMLP().to(device)
n_params = sum(p.numel() for p in mlp.parameters() if p.requires_grad)
print(f'MLP parameters: {n_params:,}')
print(mlp)
```

```python
# DataLoaders — use the same local train/val split as XGBoost
def make_loader(X, Y, shuffle=True):
    ds = TensorDataset(torch.tensor(X), torch.tensor(Y))
    return DataLoader(ds, batch_size=MLP_BATCH_SIZE, shuffle=shuffle)

train_loader = make_loader(Xtr_xgb, Ytr_xgb)
val_loader   = make_loader(Xval_xgb, Yval_xgb, shuffle=False)
test_loader  = make_loader(Xte, Yte, shuffle=False)

optimizer = torch.optim.AdamW(mlp.parameters(), lr=MLP_LR, weight_decay=MLP_WD)
scheduler = torch.optim.lr_scheduler.CosineAnnealingLR(optimizer, T_max=MLP_EPOCHS)
criterion = nn.MSELoss()

best_val_loss = float('inf')
best_state    = None
patience_cnt  = 0
train_losses, val_losses = [], []

for epoch in range(1, MLP_EPOCHS + 1):
    # Train
    mlp.train()
    total = 0.0
    for xb, yb in train_loader:
        xb, yb = xb.to(device), yb.to(device)
        optimizer.zero_grad()
        loss = criterion(mlp(xb), yb)
        loss.backward()
        optimizer.step()
        total += loss.item() * len(xb)
    train_losses.append(total / len(Xtr_xgb))

    # Validate
    mlp.eval()
    total = 0.0
    with torch.no_grad():
        for xb, yb in val_loader:
            xb, yb = xb.to(device), yb.to(device)
            total += criterion(mlp(xb), yb).item() * len(xb)
    val_loss = total / len(Xval_xgb)
    val_losses.append(val_loss)
    scheduler.step()

    if val_loss < best_val_loss:
        best_val_loss = val_loss
        best_state = {k: v.cpu().clone() for k, v in mlp.state_dict().items()}
        patience_cnt = 0
    else:
        patience_cnt += 1
        if patience_cnt >= MLP_PATIENCE:
            print(f'Early stopping at epoch {epoch}')
            break

    if epoch % 50 == 0:
        print(f'Epoch {epoch:4d}/{MLP_EPOCHS}  '
              f'train={train_losses[-1]:.4f}  val={val_loss:.4f}')

print(f'Best val MSE: {best_val_loss:.4f}')
mlp.load_state_dict(best_state)
print('Loaded best checkpoint')
```

## MLP hyperparameter search (random search)

```python

import time

# ── Search space ────────────────────────────────────────────────────────────
SEARCH_SPACE = {
    'hidden_dims': [
        [128, 128],
        [256, 256],
        [512, 512],
        [256, 256, 128],
        [512, 256, 128],
        [512, 512, 256],
        [256, 256, 256, 128],
        [512, 256, 256, 128],
    ],
    'dropout':      [0.0, 0.05, 0.1, 0.2],
    'lr':           [3e-4, 1e-3, 3e-3],
    'weight_decay': [1e-5, 1e-4, 1e-3],
}

N_TRIALS      = 30
TUNE_EPOCHS   = 200
TUNE_PATIENCE = 20

rng_search = np.random.default_rng(0)

def build_mlp(hidden_dims, dropout):
    layers = []
    in_d = N_INPUT
    for h in hidden_dims:
        layers += [nn.Linear(in_d, h), nn.GELU(), nn.LayerNorm(h), nn.Dropout(dropout)]
        in_d = h
    layers.append(nn.Linear(in_d, N_OUTPUT))
    return nn.Sequential(*layers)


def sample_cfg():
    return {
        'hidden_dims':  SEARCH_SPACE['hidden_dims'][
                            rng_search.integers(len(SEARCH_SPACE['hidden_dims']))],
        'dropout':      SEARCH_SPACE['dropout'][
                            rng_search.integers(len(SEARCH_SPACE['dropout']))],
        'lr':           SEARCH_SPACE['lr'][
                            rng_search.integers(len(SEARCH_SPACE['lr']))],
        'weight_decay': SEARCH_SPACE['weight_decay'][
                            rng_search.integers(len(SEARCH_SPACE['weight_decay']))],
    }


def train_trial(cfg):
    torch.manual_seed(SEED)
    net   = build_mlp(cfg['hidden_dims'], cfg['dropout']).to(device)
    opt   = torch.optim.AdamW(net.parameters(), lr=cfg['lr'], weight_decay=cfg['weight_decay'])
    sched = torch.optim.lr_scheduler.CosineAnnealingLR(opt, T_max=TUNE_EPOCHS)
    crit  = nn.MSELoss()

    best_val, best_ep, patience_cnt, best_state = float('inf'), 0, 0, None
    for epoch in range(1, TUNE_EPOCHS + 1):
        net.train()
        for xb, yb in train_loader:
            xb, yb = xb.to(device), yb.to(device)
            opt.zero_grad(); crit(net(xb), yb).backward(); opt.step()
        sched.step()

        net.eval()
        val_loss = sum(
            crit(net(xb.to(device)), yb.to(device)).item() * len(xb)
            for xb, yb in val_loader
        ) / len(Xval_xgb)

        if val_loss < best_val:
            best_val, best_ep, patience_cnt = val_loss, epoch, 0
            best_state = {k: v.cpu().clone() for k, v in net.state_dict().items()}
        else:
            patience_cnt += 1
            if patience_cnt >= TUNE_PATIENCE:
                break

    net.load_state_dict(best_state)
    return net, best_val, best_ep


# ── Random search ────────────────────────────────────────────────────────────
trials = []
for trial_i in range(N_TRIALS):
    cfg = sample_cfg()
    t0  = time.time()
    _, val_loss, best_ep = train_trial(cfg)
    elapsed = time.time() - t0
    trials.append({'cfg': cfg, 'val_loss': val_loss, 'best_ep': best_ep})
    print(f'Trial {trial_i+1:2d}/{N_TRIALS}  val={val_loss:.4f}  ep={best_ep:3d}  '
          f't={elapsed:.0f}s  {cfg}')

trials.sort(key=lambda x: x['val_loss'])
best_cfg = trials[0]['cfg']
print(f'\nBest config (val={trials[0]["val_loss"]:.4f}):')
print(best_cfg)
```

```python

# ── Retrain best config for longer on full train split ────────────────────
FINAL_EPOCHS   = 500
FINAL_PATIENCE = 40

torch.manual_seed(SEED)
mlp_tuned = build_mlp(best_cfg['hidden_dims'], best_cfg['dropout']).to(device)
opt_f  = torch.optim.AdamW(mlp_tuned.parameters(),
                            lr=best_cfg['lr'], weight_decay=best_cfg['weight_decay'])
sched_f = torch.optim.lr_scheduler.CosineAnnealingLR(opt_f, T_max=FINAL_EPOCHS)
crit    = nn.MSELoss()

best_val_f, patience_cnt_f = float('inf'), 0
best_state_f = None
train_losses_f, val_losses_f = [], []

for epoch in range(1, FINAL_EPOCHS + 1):
    mlp_tuned.train()
    total = 0.0
    for xb, yb in train_loader:
        xb, yb = xb.to(device), yb.to(device)
        opt_f.zero_grad(); loss = crit(mlp_tuned(xb), yb); loss.backward(); opt_f.step()
        total += loss.item() * len(xb)
    train_losses_f.append(total / len(Xtr_xgb))

    mlp_tuned.eval()
    val_loss = sum(
        crit(mlp_tuned(xb.to(device)), yb.to(device)).item() * len(xb)
        for xb, yb in val_loader
    ) / len(Xval_xgb)
    val_losses_f.append(val_loss)
    sched_f.step()

    if val_loss < best_val_f:
        best_val_f, patience_cnt_f = val_loss, 0
        best_state_f = {k: v.cpu().clone() for k, v in mlp_tuned.state_dict().items()}
    else:
        patience_cnt_f += 1
        if patience_cnt_f >= FINAL_PATIENCE:
            print(f'Early stopping at epoch {epoch}')
            break

    if epoch % 100 == 0:
        print(f'Epoch {epoch:4d}/{FINAL_EPOCHS}  '
              f'train={train_losses_f[-1]:.4f}  val={val_loss:.4f}')

mlp_tuned.load_state_dict(best_state_f)
print(f'\nBest val MSE: {best_val_f:.4f}')

# Training curve
fig, ax = plt.subplots(figsize=(8, 4))
ax.plot(train_losses_f, label='train')
ax.plot(val_losses_f,   label='val')
ax.set_xlabel('Epoch'); ax.set_ylabel('MSE (standardised)')
ax.set_title(f'MLP tuned — training curve\n{best_cfg}')
ax.legend(); ax.set_yscale('log'); ax.grid(alpha=0.3)
plt.tight_layout()
plt.savefig('fig/mlp_tuned_training_curve.png', dpi=150, bbox_inches='tight')
plt.show()
```

```python

# ── Evaluate tuned MLP ───────────────────────────────────────────────────────
mlp_tuned.eval()
preds = []
with torch.no_grad():
    for xb, _ in test_loader:
        preds.append(mlp_tuned(xb.to(device)).cpu().numpy())
mlp_tuned_preds = np.concatenate(preds, axis=0)

res_mlp_tuned = evaluate(mlp_tuned_preds, Yte, 'MLP tuned (test)')

print(f'\nMacro RMSE improvement: '
      f'{res_mlp["macro_rmse"]:.4f} → {res_mlp_tuned["macro_rmse"]:.4f}  '
      f'({(res_mlp["macro_rmse"] - res_mlp_tuned["macro_rmse"]) / res_mlp["macro_rmse"] * 100:.1f}% reduction)')

# ── Pred vs true: all 32 outputs ─────────────────────────────────────────────
from matplotlib.colors import LogNorm
from matplotlib.lines import Line2D

Y_pred_phys = unscale_Y(mlp_tuned_preds)
Y_true_phys = unscale_Y(Yte)

BLOCK_POS = {'SW': (1, 0), 'SE': (1, 1), 'NW': (0, 0), 'NE': (0, 1)}

n_rows, n_cols = 4, 8
fig, axes = plt.subplots(n_rows, n_cols, figsize=(n_cols * 3.5, n_rows * 3.5))

for c, corner in enumerate(CORNER_NAMES):
    br, bc = BLOCK_POS[corner]
    for k, mode in enumerate(ZERNIKE_MODES):
        row = br * 2 + k // 4
        col = bc * 4 + k % 4
        ax  = axes[row, col]
        t = Y_true_phys[:, c, k]
        p = Y_pred_phys[:, c, k]
        vmin = min(t.min(), p.min()); vmax = max(t.max(), p.max())
        margin = (vmax - vmin) * 0.05
        lim = (vmin - margin, vmax + margin)
        ax.hist2d(t, p, bins=50, range=[[-1, 1], [-1, 1]], cmap='magma', norm=LogNorm())
        ax.plot(lim, lim, 'b--', lw=0.7, alpha=0.7)
        ax.set_aspect('equal', adjustable='box')
        ax.set_title(f'{corner} · {mode}', fontsize=11)
        ax.tick_params(labelsize=9)
        if col == 0 or col == 4:
            ax.set_ylabel('Pred [μm]', fontsize=10)
        if row == n_rows - 1:
            ax.set_xlabel('True [μm]', fontsize=10)

for corner, (br, bc) in BLOCK_POS.items():
    cx = (bc * 4 + 1.5) / n_cols
    cy = 1.0 - br * 0.5 - 0.02
    fig.text(cx, cy, corner, ha='center', va='bottom', fontsize=14, fontweight='bold',
             transform=fig.transFigure)

fig.suptitle('MLP tuned — shared representation\nPredicted vs True (test set)', fontsize=16, y=1.03)
plt.tight_layout()

pos_l = axes[0, 3].get_position(); pos_r = axes[0, 4].get_position()
pos_t = axes[1, 0].get_position(); pos_b = axes[2, 0].get_position()
x_div = (pos_l.x1 + pos_r.x0) / 2
y_div = (pos_t.y0 + pos_b.y1) / 2
fig.add_artist(Line2D([x_div, x_div], [0.01, 0.99], transform=fig.transFigure,
                      color='white', linewidth=3, zorder=10))
fig.add_artist(Line2D([0.01, 0.99], [y_div, y_div], transform=fig.transFigure,
                      color='white', linewidth=3, zorder=10))

plt.savefig('fig/mlp_tuned_pred_vs_true_all32.png', dpi=150, bbox_inches='tight')
plt.show()
```

```python

# Load obs_scaled from dataset (re-read so this section is self-contained)
_data = np.load(DATASET_PATH)
obs_scaled = _data['obs_scaled']   # [N, 2]  zenith_distance, sky_bg  (standardised)

# Build the 4 input variants  (all standardised, using idx_train/idx_test already in scope)
VARIANTS = {
    'MLP-2nd':     X_flat[:, :N_RAFTS*3],                               # [N, 63]
    'MLP-2nd3rd':  X_flat[:, :N_RAFTS*7],                               # [N, 147]
    'MLP-all':     X_flat,                                               # [N, 168]
    'MLP-all+obs': np.concatenate([X_flat, obs_scaled], axis=1),        # [N, 170]
}

print('Input dimensions per variant:')
for name, X_v in VARIANTS.items():
    print(f'  {name}: {X_v.shape[1]}')
```

```python

def train_final_mlp(X_all, in_dim):
    """Train with best_cfg on the given input matrix. Returns (model, test_preds)."""
    Xtr_v  = X_all[idx_train][train_local].astype(np.float32)
    Xval_v = X_all[idx_train][val_local].astype(np.float32)
    Xte_v  = X_all[idx_test].astype(np.float32)

    tr_loader  = make_loader(Xtr_v,  Ytr_xgb)
    val_loader_v = make_loader(Xval_v, Yval_xgb, shuffle=False)
    te_loader  = make_loader(Xte_v,  Yte,     shuffle=False)

    # Rebuild MLP with correct in_dim but same hidden_dims / dropout from best_cfg
    layers = []
    d = in_dim
    for h in best_cfg['hidden_dims']:
        layers += [nn.Linear(d, h), nn.GELU(), nn.LayerNorm(h),
                   nn.Dropout(best_cfg['dropout'])]
        d = h
    layers.append(nn.Linear(d, N_OUTPUT))
    net = nn.Sequential(*layers).to(device)

    torch.manual_seed(SEED)
    opt   = torch.optim.AdamW(net.parameters(),
                               lr=best_cfg['lr'], weight_decay=best_cfg['weight_decay'])
    sched = torch.optim.lr_scheduler.CosineAnnealingLR(opt, T_max=FINAL_EPOCHS)
    crit  = nn.MSELoss()

    best_val, patience_cnt, best_state = float('inf'), 0, None
    for epoch in range(1, FINAL_EPOCHS + 1):
        net.train()
        for xb, yb in tr_loader:
            xb, yb = xb.to(device), yb.to(device)
            opt.zero_grad(); crit(net(xb), yb).backward(); opt.step()
        sched.step()

        net.eval()
        val_loss = sum(
            crit(net(xb.to(device)), yb.to(device)).item() * len(xb)
            for xb, yb in val_loader_v
        ) / len(Xval_v)

        if val_loss < best_val:
            best_val, patience_cnt = val_loss, 0
            best_state = {k: v.cpu().clone() for k, v in net.state_dict().items()}
        else:
            patience_cnt += 1
            if patience_cnt >= FINAL_PATIENCE:
                break

    net.load_state_dict(best_state)

    net.eval()
    preds = []
    with torch.no_grad():
        for xb, _ in te_loader:
            preds.append(net(xb.to(device)).cpu().numpy())
    return net, np.concatenate(preds, axis=0), best_val


# Train all 4 variants

```

```python
ablation_results = {}

for name, X_v in VARIANTS.items():
    print(f'\nTraining {name}  (in_dim={X_v.shape[1]}) ...')
    net, preds, bval = train_final_mlp(X_v, X_v.shape[1])
    res = evaluate(preds, Yte, f'{name} (test)')
    ablation_results[name] = {'res': res, 'preds': preds, 'model': net, 'best_val': bval}
```

```python

# Pred vs true all-32 plot for each ablation variant
from matplotlib.colors import LogNorm
from matplotlib.lines import Line2D

BLOCK_POS = {'SW': (1, 0), 'SE': (1, 1), 'NW': (0, 0), 'NE': (0, 1)}

for name, d in ablation_results.items():
    Y_pred_phys = unscale_Y(d['preds'])
    Y_true_phys = unscale_Y(Yte)

    n_rows, n_cols = 4, 8
    fig, axes = plt.subplots(n_rows, n_cols, figsize=(n_cols * 3.5, n_rows * 3.5))

    for c, corner in enumerate(CORNER_NAMES):
        br, bc = BLOCK_POS[corner]
        for k, mode in enumerate(ZERNIKE_MODES):
            row = br * 2 + k // 4
            col = bc * 4 + k % 4
            ax  = axes[row, col]
            t = Y_true_phys[:, c, k]
            p = Y_pred_phys[:, c, k]
            vmin = min(t.min(), p.min()); vmax = max(t.max(), p.max())
            margin = (vmax - vmin) * 0.05
            lim = (vmin - margin, vmax + margin)
            ax.hist2d(t, p, bins=50, range=[[-1, 1], [-1, 1]], cmap='magma', norm=LogNorm())
            ax.plot(lim, lim, 'b--', lw=0.7, alpha=0.7)
            ax.set_aspect('equal', adjustable='box')
            ax.set_title(f'{corner} · {mode}', fontsize=11)
            ax.tick_params(labelsize=9)
            if col == 0 or col == 4:
                ax.set_ylabel('Pred [μm]', fontsize=10)
            if row == n_rows - 1:
                ax.set_xlabel('True [μm]', fontsize=10)

    for corner, (br, bc) in BLOCK_POS.items():
        cx = (bc * 4 + 1.5) / n_cols
        cy = 1.0 - br * 0.5 - 0.02
        fig.text(cx, cy, corner, ha='center', va='bottom', fontsize=14, fontweight='bold',
                 transform=fig.transFigure)

    macro = d['res']['macro_rmse']
    fig.suptitle(f'{name} — Predicted vs True (test set)  [macro RMSE={macro:.4f} μm]',
                 fontsize=16, y=1.03)
    plt.tight_layout()

    pos_l = axes[0, 3].get_position(); pos_r = axes[0, 4].get_position()
    pos_t = axes[1, 0].get_position(); pos_b = axes[2, 0].get_position()
    x_div = (pos_l.x1 + pos_r.x0) / 2
    y_div = (pos_t.y0 + pos_b.y1) / 2
    fig.add_artist(Line2D([x_div, x_div], [0.01, 0.99], transform=fig.transFigure,
                          color='white', linewidth=3, zorder=10))
    fig.add_artist(Line2D([0.01, 0.99], [y_div, y_div], transform=fig.transFigure,
                          color='white', linewidth=3, zorder=10))

    fname = name.replace('+', '_plus_').replace('-', '_')
    plt.savefig(f'fig/mlp_ablation_{fname}_pred_vs_true.png', dpi=150, bbox_inches='tight')
    plt.show()
```
