# GNN: PSF Moments → Zernike Coefficients

Trains a Graph Neural Network (GraphSAGE + corner attention readout) to predict
Zernike wavefront coefficients at the 4 AOS corners from per-raft PSF moments
measured across the LSST focal plane.

Architecture follows `gnn_guide.txt`:
- **Nodes**: 21 science rafts, features = [T, e1, e2, c11, c12, c31, c32, rho4] + (x,y) coords → dim 10
- **Edges**: side + diagonal neighbours + self-loops
- **Encoder**: 3-layer GraphSAGE (10→32→64→64)
- **Readout**: corner attention (4 corners)
- **Decoder**: MLP per corner (128→64→32→8 Zernike modes)
- **Output**: Y ∈ ℝ^{B×4×8}

```python
import numpy as np
import pandas as pd  # kept for potential inspection
import pickle
import torch
import torch.nn as nn
import torch.nn.functional as F
from torch_geometric.data import Data
from torch_geometric.loader import DataLoader
from torch_geometric.nn import SAGEConv, global_mean_pool
import matplotlib.pyplot as plt

device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print(f'Using device: {device}')
```

## Config

```python
MODEL_PATH = 'gnn_psf_zernike.pt'

SCIENCE_RAFTS = [
    'R01', 'R02', 'R03',
    'R10', 'R11', 'R12', 'R13', 'R14',
    'R20', 'R21', 'R22', 'R23', 'R24',
    'R30', 'R31', 'R32', 'R33', 'R34',
    'R41', 'R42', 'R43',
]
N_RAFTS = len(SCIENCE_RAFTS)  # 21

RAFT_GRID = [
    [None,  'R01', 'R02', 'R03', None ],
    ['R10', 'R11', 'R12', 'R13', 'R14'],
    ['R20', 'R21', 'R22', 'R23', 'R24'],
    ['R30', 'R31', 'R32', 'R33', 'R34'],
    [None,  'R41', 'R42', 'R43', None ],
]

MOMENT_KEYS  = ['T', 'e1', 'e2', 'c11', 'c12', 'c31', 'c32', 'rho4']
N_MOMENTS    = len(MOMENT_KEYS)   # 8

ZERNIKE_MODES = ['z4', 'z5', 'z6', 'z7', 'z8', 'z9', 'z10', 'z11']
N_ZERNIKE     = len(ZERNIKE_MODES)  # 8

# AOS wavefront sensor detectors per corner (SW, SE, NW, NE)
CORNER_DETECTORS = [191, 195, 199, 203]
CORNER_NAMES     = ['SW', 'SE', 'NW', 'NE']
N_CORNERS        = 4

# Normalised focal-plane coordinates for corner attention
CORNER_COORDS = [[-1., -1.], [1., -1.], [-1., 1.], [1., 1.]]

# Model / training hyper-params
HIDDEN_DIM  = 64
IN_CHANNELS = N_MOMENTS + 2   # 8 PSF features + 2 raft coords = 10
BATCH_SIZE  = 16
N_EPOCHS    = 300
LR          = 1e-3
TEST_FRAC   = 0.3
SEED        = 42
```

## Graph structure

```python
raft_to_idx = {r: i for i, r in enumerate(SCIENCE_RAFTS)}

# --- Edge index (side + diagonal neighbours + self-loops) ---
src_list, dst_list = [], []
for r, row in enumerate(RAFT_GRID):
    for c, raft in enumerate(row):
        if raft is None:
            continue
        for dr in [-1, 0, 1]:
            for dc in [-1, 0, 1]:
                if dr == 0 and dc == 0:
                    continue
                nr, nc = r + dr, c + dc
                if 0 <= nr < 5 and 0 <= nc < 5:
                    neighbour = RAFT_GRID[nr][nc]
                    if neighbour is not None:
                        src_list.append(raft_to_idx[raft])
                        dst_list.append(raft_to_idx[neighbour])

# Self-loops
for i in range(N_RAFTS):
    src_list.append(i)
    dst_list.append(i)

edge_index = torch.tensor([src_list, dst_list], dtype=torch.long)
print(f'Graph: {N_RAFTS} nodes, {edge_index.size(1)} edges')

# --- Raft coordinates (normalised grid position) ---
raft_coords = np.zeros((N_RAFTS, 2), dtype=np.float32)
for r, row in enumerate(RAFT_GRID):
    for c, raft in enumerate(row):
        if raft is not None:
            raft_coords[raft_to_idx[raft]] = [(c - 2) / 2.0, (r - 2) / 2.0]

raft_coords_t = torch.tensor(raft_coords)  # [21, 2]
```

## Load preprocessed dataset

```python
# Load preprocessed arrays and scalers produced by prepare_dataset.ipynb
data = np.load('data/dataset.npz')
X_scaled  = data['X_scaled']   # [N, 21, 8] standardised
Y_scaled  = data['Y_scaled']   # [N, 4, 8]  standardised
X_raw     = data['X_raw']
Y_raw     = data['Y_raw']
idx_train = data['idx_train']
idx_test  = data['idx_test']
N = len(X_scaled)

with open('data/scalers.pkl', 'rb') as f:
    scalers = pickle.load(f)
x_scaler = scalers['x_scaler']
y_scaler = scalers['y_scaler']

def unscale_Y(Y_s):
    shape = Y_s.shape
    return y_scaler.inverse_transform(Y_s.reshape(-1, N_ZERNIKE)).reshape(shape)

print(f'Train: {len(idx_train)}  Test: {len(idx_test)}  Total: {N}')
```

## Build torch_geometric dataset

```python
def make_dataset(indices):
    dataset = []
    for i in indices:
        x_feat = torch.tensor(X_scaled[i])        # [21, 8]
        x_node = torch.cat([x_feat, raft_coords_t], dim=-1)  # [21, 10]
        y = torch.tensor(Y_scaled[i])              # [4, 8]
        dataset.append(Data(x=x_node, edge_index=edge_index, y=y))
    return dataset

train_dataset = make_dataset(idx_train)
test_dataset  = make_dataset(idx_test)

train_loader = DataLoader(train_dataset, batch_size=BATCH_SIZE, shuffle=True)
test_loader  = DataLoader(test_dataset,  batch_size=BATCH_SIZE, shuffle=False)
print(f'Train batches: {len(train_loader)}  Test batches: {len(test_loader)}')
```

## Model definition

```python
class CornerAttention(nn.Module):
    """Attention readout that produces one embedding per focal-plane corner."""

    def __init__(self, hidden_dim, corner_coords):
        super().__init__()
        self.register_buffer(
            'corner_coords', torch.tensor(corner_coords, dtype=torch.float)
        )  # [4, 2]
        self.coord_embed = nn.Sequential(
            nn.Linear(2, hidden_dim),
            nn.GELU(),
            nn.Linear(hidden_dim, hidden_dim),
        )  # q_c = MLP(corner_coord)
        self.W = nn.Linear(hidden_dim, hidden_dim, bias=False)

    def forward(self, H, batch):
        # H: [B*N_RAFTS, D],  batch: [B*N_RAFTS]
        B = int(batch.max().item()) + 1
        D = H.size(-1)

        H_b  = H.view(B, N_RAFTS, D)              # [B, 21, D]
        Wh   = self.W(H_b)                         # [B, 21, D]
        q    = self.coord_embed(self.corner_coords) # [4, D]

        scores = torch.einsum('cd,bnd->bcn', q, Wh)  # [B, 4, 21]
        attn   = F.softmax(scores, dim=-1)            # [B, 4, 21]
        h_c    = torch.einsum('bcn,bnd->bcd', attn, H_b)  # [B, 4, D]
        return h_c, attn


class PSFToZernikeGNN(nn.Module):
    def __init__(self, in_channels=IN_CHANNELS, hidden_dim=HIDDEN_DIM,
                 n_zernike=N_ZERNIKE, corner_coords=CORNER_COORDS):
        super().__init__()

        # --- GNN encoder (10 → 32 → 64 → 64) ---
        self.conv1 = SAGEConv(in_channels, 32)
        self.norm1 = nn.LayerNorm(32)
        self.conv2 = SAGEConv(32, hidden_dim)
        self.norm2 = nn.LayerNorm(hidden_dim)
        self.conv3 = SAGEConv(hidden_dim, hidden_dim)
        self.norm3 = nn.LayerNorm(hidden_dim)
        self.drop  = nn.Dropout(0.1)

        # --- Corner attention readout ---
        self.corner_attn = CornerAttention(hidden_dim, corner_coords)

        # --- Per-corner decoder MLP (128 → 64 → 32 → n_zernike) ---
        self.decoder = nn.Sequential(
            nn.Linear(hidden_dim * 2, 64),
            nn.GELU(),
            nn.Linear(64, 32),
            nn.GELU(),
            nn.Linear(32, n_zernike),
        )

    def forward(self, x, edge_index, batch):
        # Encoder
        x = self.drop(F.gelu(self.norm1(self.conv1(x, edge_index))))
        x = self.drop(F.gelu(self.norm2(self.conv2(x, edge_index))))
        x = self.drop(F.gelu(self.norm3(self.conv3(x, edge_index))))
        H = x  # [B*21, 64]

        # Global pooling
        h_global = global_mean_pool(H, batch)   # [B, 64]

        # Corner attention
        h_corners, attn = self.corner_attn(H, batch)   # [B, 4, 64]

        # Decoder
        h_global_exp = h_global.unsqueeze(1).expand(-1, N_CORNERS, -1)  # [B, 4, 64]
        h_cat = torch.cat([h_global_exp, h_corners], dim=-1)            # [B, 4, 128]
        return self.decoder(h_cat)   # [B, 4, n_zernike]


model = PSFToZernikeGNN().to(device)
n_params = sum(p.numel() for p in model.parameters() if p.requires_grad)
print(f'Parameters: {n_params:,}')
print(model)
```

## Training

```python
optimizer = torch.optim.Adam(model.parameters(), lr=LR)
criterion = nn.MSELoss()

train_losses, test_losses = [], []

for epoch in range(1, N_EPOCHS + 1):
    # --- Train ---
    model.train()
    total = 0.0
    for batch in train_loader:
        batch = batch.to(device)
        B = batch.num_graphs
        optimizer.zero_grad()
        pred = model(batch.x, batch.edge_index, batch.batch)          # [B, 4, 8]
        target = batch.y.view(B, N_CORNERS, N_ZERNIKE)                # [B, 4, 8]
        loss = criterion(pred, target)
        loss.backward()
        optimizer.step()
        total += loss.item() * B
    train_losses.append(total / len(train_dataset))

    # --- Evaluate ---
    model.eval()
    total = 0.0
    with torch.no_grad():
        for batch in test_loader:
            batch = batch.to(device)
            B = batch.num_graphs
            pred   = model(batch.x, batch.edge_index, batch.batch)
            target = batch.y.view(B, N_CORNERS, N_ZERNIKE)
            total += criterion(pred, target).item() * B
    test_losses.append(total / len(test_dataset))

    if epoch % 10 == 0:
        print(f'Epoch {epoch:4d}/{N_EPOCHS}  '
              f'train MSE={train_losses[-1]:.4f}  '
              f'test MSE={test_losses[-1]:.4f}')
```

```python
fig, ax = plt.subplots(figsize=(8, 4))
ax.plot(train_losses, label='train')
ax.plot(test_losses,  label='test')
ax.set_xlabel('Epoch')
ax.set_ylabel('MSE (standardised)')
ax.set_title('Training curve')
ax.legend()
ax.set_yscale('log')
ax.grid(alpha=0.3)
plt.tight_layout()
plt.savefig('gnn_training_curve.png', dpi=150, bbox_inches='tight')
plt.show()
```

## Save model + scalers

```python
torch.save({
    'model_state_dict': model.state_dict(),
    'x_scaler': x_scaler,
    'y_scaler': y_scaler,
    'config': {
        'in_channels': IN_CHANNELS,
        'hidden_dim':  HIDDEN_DIM,
        'n_zernike':   N_ZERNIKE,
        'corner_coords': CORNER_COORDS,
    },
    'edge_index': edge_index,
    'raft_coords': raft_coords,
}, MODEL_PATH)
print(f'Saved to {MODEL_PATH}')
```

## Evaluate on test set

```python
model.eval()
all_pred, all_true = [], []

with torch.no_grad():
    for batch in test_loader:
        batch = batch.to(device)
        B = batch.num_graphs
        pred = model(batch.x, batch.edge_index, batch.batch)   # [B, 4, 8]
        all_pred.append(pred.cpu().numpy())
        all_true.append(batch.y.view(B, N_CORNERS, N_ZERNIKE).cpu().numpy())

Y_pred_s = np.concatenate(all_pred, axis=0)  # [N_test, 4, 8] standardised
Y_true_s = np.concatenate(all_true, axis=0)

# Inverse-transform to physical units
Y_pred = unscale_Y(Y_pred_s)  # [N_test, 4, 8]
Y_true = unscale_Y(Y_true_s)

# RMSE per Zernike mode (averaged over corners and samples)
rmse_mode = np.sqrt(np.mean((Y_pred - Y_true)**2, axis=(0, 1)))  # [8]
print('RMSE per Zernike mode [μm]:')
for mode, rmse in zip(ZERNIKE_MODES, rmse_mode):
    print(f'  {mode}: {rmse:.4f}')

# RMSE per corner (averaged over modes and samples)
rmse_corner = np.sqrt(np.mean((Y_pred - Y_true)**2, axis=(0, 2)))  # [4]
print('\nRMSE per corner [μm]:')
for corner, rmse in zip(CORNER_NAMES, rmse_corner):
    print(f'  {corner}: {rmse:.4f}')

# Pearson r per mode
print('\nPearson r per Zernike mode:')
for k, mode in enumerate(ZERNIKE_MODES):
    p = Y_pred[:, :, k].ravel()
    t = Y_true[:, :, k].ravel()
    r = np.corrcoef(p, t)[0, 1]
    print(f'  {mode}: {r:.3f}')
```

```python

# Predicted vs true: all 32 outputs — GNN
# Layout: 4 rows × 8 cols — corners at focal-plane positions
#   NW (top-left)  | NE (top-right)
#   SW (bot-left)  | SE (bot-right)
import os
from matplotlib.colors import LogNorm
from matplotlib.lines import Line2D

os.makedirs('fig', exist_ok=True)

BLOCK_POS = {'SW': (1, 0), 'SE': (1, 1), 'NW': (0, 0), 'NE': (0, 1)}

n_rows, n_cols = 4, 8
fig, axes = plt.subplots(n_rows, n_cols, figsize=(n_cols * 3.5, n_rows * 3.5))

for c, corner in enumerate(CORNER_NAMES):
    br, bc = BLOCK_POS[corner]
    for k, mode in enumerate(ZERNIKE_MODES):
        row = br * 2 + k // 4
        col = bc * 4 + k % 4
        ax  = axes[row, col]
        t = Y_true[:, c, k]
        p = Y_pred[:, c, k]
        vmin = min(t.min(), p.min()); vmax = max(t.max(), p.max())
        margin = (vmax - vmin) * 0.05
        lim = (vmin - margin, vmax + margin)
        ax.hist2d(t, p, bins=50, range=[[-1, 1], [-1, 1]], cmap='magma', norm=LogNorm())
        ax.plot(lim, lim, 'b--', lw=0.7, alpha=0.7)
        ax.set_aspect('equal', adjustable='box')
        ax.set_title(f'{corner} · {mode}', fontsize=11)
        ax.tick_params(labelsize=9)
        if col == 0 or col == 4:
            ax.set_ylabel('Pred [μm]', fontsize=10)
        if row == n_rows - 1:
            ax.set_xlabel('True [μm]', fontsize=10)

for corner, (br, bc) in BLOCK_POS.items():
    cx = (bc * 4 + 1.5) / n_cols
    cy = 1.0 - br * 0.5 - 0.02
    fig.text(cx, cy, corner, ha='center', va='bottom', fontsize=14, fontweight='bold',
             transform=fig.transFigure)

fig.suptitle('GNN (GraphSAGE + corner attention)\nPredicted vs True (test set)', fontsize=16, y=1.03)
plt.tight_layout()

pos_l = axes[0, 3].get_position(); pos_r = axes[0, 4].get_position()
pos_t = axes[1, 0].get_position(); pos_b = axes[2, 0].get_position()
x_div = (pos_l.x1 + pos_r.x0) / 2
y_div = (pos_t.y0 + pos_b.y1) / 2
fig.add_artist(Line2D([x_div, x_div], [0.01, 0.99], transform=fig.transFigure,
                      color='white', linewidth=3, zorder=10))
fig.add_artist(Line2D([0.01, 0.99], [y_div, y_div], transform=fig.transFigure,
                      color='white', linewidth=3, zorder=10))

plt.savefig('fig/gnn_pred_vs_true_all32.png', dpi=150, bbox_inches='tight')
plt.show()
```

```python
# Predicted vs true scatter for each Zernike mode
ncols = 4
nrows = int(np.ceil(N_ZERNIKE / ncols))
fig, axes = plt.subplots(nrows, ncols, figsize=(14, 6))
axes = axes.ravel()

for k, mode in enumerate(ZERNIKE_MODES):
    ax = axes[k]
    t = Y_true[:, :, k].ravel()
    p = Y_pred[:, :, k].ravel()
    ax.scatter(t, p, s=8, alpha=0.5)
    lims = [min(t.min(), p.min()), max(t.max(), p.max())]
    ax.plot(lims, lims, 'r--', lw=0.8)
    r = np.corrcoef(t, p)[0, 1]
    ax.set_title(f'{mode}  r={r:.2f}', fontsize=10)
    ax.set_xlabel('True [μm]', fontsize=8)
    ax.set_ylabel('Pred [μm]', fontsize=8)
    ax.grid(alpha=0.3)

for ax in axes[N_ZERNIKE:]:
    ax.set_visible(False)

plt.suptitle('Predicted vs True Zernike (test set)', fontsize=12)
plt.tight_layout()
plt.savefig('gnn_pred_vs_true.png', dpi=150, bbox_inches='tight')
plt.show()
```

```python
# RMSE bar chart: per mode and per corner
fig, axes = plt.subplots(1, 2, figsize=(12, 4))

axes[0].bar(ZERNIKE_MODES, rmse_mode)
axes[0].set_xlabel('Zernike mode')
axes[0].set_ylabel('RMSE [μm]')
axes[0].set_title('RMSE per Zernike mode')
axes[0].grid(alpha=0.3, axis='y')

axes[1].bar(CORNER_NAMES, rmse_corner)
axes[1].set_xlabel('Corner')
axes[1].set_ylabel('RMSE [μm]')
axes[1].set_title('RMSE per corner')
axes[1].grid(alpha=0.3, axis='y')

plt.tight_layout()
plt.savefig('gnn_rmse.png', dpi=150, bbox_inches='tight')
plt.show()
```
